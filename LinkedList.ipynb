```python
# search through traversing
class listnode:
  def __init__(self, val):
    self.val = val
    self.next = None
head = listnode(10)
head.next = listnode(20)
head.next.next = listnode(30)
head.next.next.next = listnode(40)
head.next.next.next.next = listnode(50)
# traverse
temp = head
found = False
num = int(input("input number to find"))
while temp:
  if temp.val == num:
    found = True
    print("found", num)
    break
  temp = temp.next
if not found:
  print("not found")


```

```python
#insertion at the beginning
class listnode:
  def __init__(self, val):
    self.val = val
    self.next = None
head = listnode(10)
head.next = listnode(20)
head.next.next = listnode(30)
head.next.next.next = listnode(40)
head.next.next.next.next = listnode(50)

# insertion logic
new = listnode(5)
new.next = head
head = new

# traverse
temp = head
while temp:
  print(temp.val)
  temp = temp.next
```

```python
# inserting at mid using 2 ptr fast , slow . this track the mid , and mid gets
# exact ceter value if there are odd number of nodes.
class ll:
  def __init__(self, val):
    self.val = val
    self.next = None
head = ll(10)
head.next = ll(20)
head.next.next = ll(30)
head.next.next.next = ll(40)
head.next.next.next.next = ll(50)


# 2 ptr
fast = head
slow = head
prev = None
while fast and fast.next:
  prev = slow
  slow = slow.next
  fast = fast.next.next
new = ll(35)
new.next = slow.next
slow.next = new

# view
temp = head
while temp:
  print(temp.val)
  temp = temp.next




```

```python
# inserting at mid using 3 ptr - fast , slow , prev .  prev track the node just
# before slow , because in even number there are 2 mids , and slow points to the
# 2nd mid .
class ll:
  def __init__(self, val):
    self.val = val
    self.next = None
head = ll(10)
head.next = ll(20)
head.next.next = ll(30)
head.next.next.next = ll(40)
head.next.next.next.next = ll(50)
head.next.next.next.next.next = ll(60)

# 2 ptr
fast = head
slow = head
prev = None
while fast and fast.next:
  prev = slow
  slow = slow.next
  fast = fast.next.next
new = ll(35)
new.next = prev.next
prev.next = new

# view
temp = head
while temp:
  print(temp.val)
  temp = temp.next




```
